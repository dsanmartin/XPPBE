## Tutorial jupyter notebook

#### Specify the simulation inputs

```python
import os
from xppbe.Simulation import Simulation

# Create simulation object
simulation = Simulation(yaml_path='tutorial_parameters.yaml', molecule_dir=os.getcwd())

simulation.domain_properties
```

#### Create the simulation

```python
simulation.create_simulation()
simulation.adapt_model()
```

#### Solve the model

```python
simulation.solve_model()
```

#### Postprocessing object

```python
Post = simulation.postprocessing(jupyter=True)
```

```python
Post.plot_loss_history(plot_w=False)
```

```python
Post.plot_loss_history(domain=1);
Post.plot_loss_history(domain=2);
```

```python
Post.plot_G_solv_history();
```

```python
import numpy as np
Post.plot_phi_contour(N=100, x0=np.array([0,0,0]), n=np.array([1,0,0]), value='react');
```

```python
Post.plot_interface_3D(value='react',jupyter=True);
```
